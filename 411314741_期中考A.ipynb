````python
# ============================================================
# 第 2 格：建立資料庫與初始資料 
# ==========================================
import sqlite3

def init_db():
    # 每次執行都先 DROP TABLE IF EXISTS，再重建資料表並放入初始資料
    conn = sqlite3.connect('activity_system.db')
    cursor = conn.cursor()
    
    cursor.execute("DROP TABLE IF EXISTS registrations")
    cursor.execute("DROP TABLE IF EXISTS activities")
    
    # 建立活動資料表
    cursor.execute('''
        CREATE TABLE activities (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT UNIQUE,
            date TEXT,
            quota INTEGER
        )
    ''')
    
    # 建立報名紀錄資料表
    cursor.execute('''
        CREATE TABLE registrations (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT,
            activity_name TEXT,
            headcount INTEGER,
            status TEXT
        )
    ''')
    
    # 放入初始資料
    initial_activities = [
        ('迎新晚會', '2026-11-20', 6),
        ('程式工作坊', '2026-11-25', 4),
        ('籃球友誼賽', '2026-11-28', 3)
    ]
    cursor.executemany('INSERT INTO activities (name, date, quota) VALUES (?, ?, ?)', initial_activities)
    
    conn.commit()
    conn.close()
    print("✅ 資料庫與初始資料 建立完成！")

init_db()
````

````python
# ============================================================
# 第 6 格：功能函式
# ============================================================
# import sqlite3
import pandas as pd

def process_business_logic(student_name, user_input):
    if not student_name or not student_name.strip():
        return "⚠️ 請輸入學生姓名！"
        
    parsed = parse_user_input(user_input)
    action = parsed["action"]
    activity = parsed["activity"]
    headcount = parsed["headcount"]
    
    conn = sqlite3.connect('activity_system.db')
    cursor = conn.cursor()
    
    try:
        # [A3] 統計查詢要使用 JOIN 與 GROUP BY
        if action == "統計" or "統計" in user_input or "狀況" in user_input:
            query = '''
                SELECT 
                    a.date AS 日期,
                    a.name AS 活動名稱,
                    a.quota AS 總名額,
                    COALESCE(SUM(CASE WHEN r.status = '成功' THEN r.headcount ELSE 0 END), 0) AS 已報名人數,
                    (a.quota - COALESCE(SUM(CASE WHEN r.status = '成功' THEN r.headcount ELSE 0 END), 0)) AS 剩餘名額
                FROM activities a
                LEFT JOIN registrations r ON a.name = r.activity_name
                GROUP BY a.id
            '''
            df = pd.read_sql_query(query, conn)
            return f"【所有活動報名統計】\n\n{df.to_string(index=False)}"

        if action == "未知":
            return "🤖 無法判斷您的意圖。請嘗試輸入：「我要報名迎新晚會」、「取消程式工作坊」或「查詢籃球友誼賽」。"

        # 查詢功能
        if action == "查詢":
            if activity:
                cursor.execute('''
                    SELECT a.quota - COALESCE(SUM(CASE WHEN r.status = '成功' THEN r.headcount ELSE 0 END), 0)
                    FROM activities a
                    LEFT JOIN registrations r ON a.name = r.activity_name AND r.status = '成功'
                    WHERE a.name = ?
                    GROUP BY a.id
                ''', (activity,))
                row = cursor.fetchone()
                # fetchone() 之後一定要先檢查是否為 None
                if row is None:
                    return f"❌ 找不到活動【{activity}】。"
                return f"📋 活動【{activity}】目前剩餘名額為：{row[0]} 人。"
            else:
                return "📋 目前系學會舉辦的活動包含：迎新晚會、程式工作坊、籃球友誼賽。"

        # 報名功能
        if action == "報名":
            if not activity:
                return "❌ 報名失敗：找不到您想報名的活動名稱。"
                
            # [A1] 檢查人數是否在 1~3 之間
            if headcount < 1 or headcount > 3:
                return f"❌ 報名失敗：單次報名人數必須在 1 至 3 人之間（您輸入為 {headcount} 人）。"
                
            # [A2] 檢查每位學生最多同時報名 2 個活動
            cursor.execute('''
                SELECT COUNT(DISTINCT activity_name) FROM registrations 
                WHERE student_name = ? AND status = '成功'
            ''', (student_name,))
            row_cnt = cursor.fetchone()
            registered_count = row_cnt[0] if row_cnt else 0
            if registered_count >= 2:
                return f"❌ 報名失敗：【{student_name}】您已達到每人最多同時報名 2 個活動的上限！"
                
            # 檢查是否重複報名同一活動
            cursor.execute('''
                SELECT COUNT(*) FROM registrations 
                WHERE student_name = ? AND activity_name = ? AND status = '成功'
            ''', (student_name, activity))
            row_dup = cursor.fetchone()
            if row_dup and row_dup[0] > 0:
                return f"❌ 報名失敗：【{student_name}】您已經報名過【{activity}】了，不可重複報名！"
                
            # 檢查活動剩餘名額
            cursor.execute('''
                SELECT a.quota - COALESCE(SUM(CASE WHEN r.status = '成功' THEN r.headcount ELSE 0 END), 0)
                FROM activities a
                LEFT JOIN registrations r ON a.name = r.activity_name AND r.status = '成功'
                WHERE a.name = ?
                GROUP BY a.id
            ''', (activity,))
            row_quota = cursor.fetchone()
            if row_quota is None:
                return f"❌ 報名失敗：找不到活動【{activity}】。"
                
            remaining = row_quota[0]
            if remaining < headcount:
                return f"❌ 報名失敗：【{activity}】剩餘名額不足（剩餘 {remaining} 人，欲報名 {headcount} 人）。"
                
            # 執行寫入資料庫
            cursor.execute('''
                INSERT INTO registrations (student_name, activity_name, headcount, status)
                VALUES (?, ?, ?, '成功')
            ''', (student_name, activity, headcount))
            conn.commit()
            return f"✅ 成功！已為【{student_name}】完成報名【{activity}】{headcount} 位。"

        # 取消功能
        if action == "取消":
            if not activity:
                return "❌ 取消失敗：請指定要取消的活動名稱。"
                
            cursor.execute('''
                SELECT id FROM registrations 
                WHERE student_name = ? AND activity_name = ? AND status = '成功'
            ''', (student_name, activity))
            row_reg = cursor.fetchone()
            if row_reg is None:
                return f"❌ 取消失敗：查無【{student_name}】報名【{activity}】的有效紀錄。"
                
            reg_id = row_reg[0]
            cursor.execute("UPDATE registrations SET status = '已取消' WHERE id = ?", (reg_id,))
            conn.commit()
            
            # 查詢取消後的剩餘名額
            cursor.execute('''
                SELECT a.quota - COALESCE(SUM(CASE WHEN r.status = '成功' THEN r.headcount ELSE 0 END), 0)
                FROM activities a
                LEFT JOIN registrations r ON a.name = r.activity_name AND r.status = '成功'
                WHERE a.name = ?
                GROUP BY a.id
            ''', (activity,))
            row_rem = cursor.fetchone()
            rem_after = row_rem[0] if row_rem else 0
            
            return f"✅ 已成功為【{student_name}】取消【{activity}】。目前該活動剩餘名額：{rem_after} 人。"

        return "⚠️ 未知的操作指令。"

    except Exception as e:
        return f"⚠️ 系統 發生錯誤: {str(e)}"
    finally:
        # 資料庫連用 try/finally 確保關閉
        conn.close()

print("✅ 功能函式載入完成！")

````
