````python
# ==========================================
# 第 2 格：建立資料庫與初始資料
# ==========================================
import sqlite3

def init_db():
    conn = sqlite3.connect('activity_system.db')
    cursor = conn.cursor()

    # 每次執行都先清除舊表
    cursor.execute("DROP TABLE IF EXISTS registrations")
    cursor.execute("DROP TABLE IF EXISTS activities")

    # 建立活動資料表
    cursor.execute('''
        CREATE TABLE activities (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT UNIQUE,
            date TEXT,
            quota INTEGER
        )
    ''')

    # 建立報名紀錄資料表
    cursor.execute('''
        CREATE TABLE registrations (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT,
            activity_name TEXT,
            headcount INTEGER,
            status TEXT
        )
    ''')

    # 放入初始資料
    initial_activities = [
        ('迎新晚會', '2026-11-20', 6),
        ('程式工作坊', '2026-11-25', 4),
        ('籃球友誼賽', '2026-11-28', 3)
    ]
    cursor.executemany('INSERT INTO activities (name, date, quota) VALUES (?, ?, ?)', initial_activities)

    conn.commit()
    conn.close()
    print("✅ 資料庫與初始資料 建立完成！")

# 執行初始化函式
init_db()
````

````python
import json
import re

ai_cache = {}

def parse_user_input(text):
    if text in ai_cache:
        return ai_cache[text]

    prompt = f"""
    你是系學會活動報名系統的語意解析助理。請從下列使用者說的話中萃取資訊：
    使用者輸入：「{text}」

    規則：
    1. action 只能是：報名、取消、查詢、統計 之一。如果無法判斷請填「未知」。
    2. activity 只能是：迎新晚會、程式工作坊、籃球友誼賽 之一。如果沒有提到具體活動則填 null。
    3. headcount：尋找句子中的人數。如果沒提到預設為 1.
    請嚴格以 JSON 格式回傳：
    {{"action": "...", "activity": "...", "headcount": 1}}
    """

    response_text = ""
    for attempt in range(3):
        try:
            completion = client.chat.completions.create(
                model=MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )
            response_text = completion.choices[0].message.content
            time.sleep(2)
            break
        except Exception as e:
            if "429" in str(e) and attempt < 2:
                time.sleep(30)
            else:
                break

    try:
        clean_json = re.sub(r'```json|```', '', response_text).strip()
        data = json.loads(clean_json)
        action = data.get("action", "未知")
        activity = data.get("activity", None)

        num_match = re.search(r'\d+', str(text))
        headcount = int(num_match.group()) if num_match else int(data.get("headcount", 1))
        result = {"raw_response": response_text, "action": action, "activity": activity, "headcount": headcount}
    except Exception:
        result = {"raw_response": response_text, "action": "未知", "activity": None, "headcount": 1}

    ai_cache[text] = result
    return result

print("✅ AI 解讀函式載入完成！")
````

````python
import sqlite3
import pandas as pd

def process_business_logic(student_name, user_input):
    if not student_name or not student_name.strip():
        return "⚠️ 請輸入學生姓名！"

    parsed = parse_user_input(user_input)
    action, activity, headcount = parsed["action"], parsed["activity"], parsed["headcount"]

    conn = sqlite3.connect('activity_system.db')
    cursor = conn.cursor()

    try:
        if action == "統計" or "統計" in user_input or "狀況" in user_input:
            query = '''
                SELECT
                    a.date AS 日期, a.name AS 活動名稱, a.quota AS 總名額,
                    COALESCE(SUM(CASE WHEN r.status = '成功' THEN r.headcount ELSE 0 END), 0) AS 已報名人數,
                    (a.quota - COALESCE(SUM(CASE WHEN r.status = '成功' THEN r.headcount ELSE 0 END), 0)) AS 剩餘名額
                FROM activities a LEFT JOIN registrations r ON a.name = r.activity_name
                GROUP BY a.id
            '''
            return f"【所有活動報名統計】\n\n{pd.read_sql_query(query, conn).to_string(index=False)}"

        if action == "未知":
            return "🤖 無法判斷您的意圖。請嘗試輸入：「我要報名迎新晚會」、「取消程式工作坊」等。"

        if action == "查詢":
            if activity:
                cursor.execute('''
                    SELECT a.quota - COALESCE(SUM(CASE WHEN r.status = '成功' THEN r.headcount ELSE 0 END), 0)
                    FROM activities a LEFT JOIN registrations r ON a.name = r.activity_name AND r.status = '成功'
                    WHERE a.name = ? GROUP BY a.id
                ''', (activity,))
                row = cursor.fetchone()
                if row is None: return f"❌ 找不到活動【{activity}】。"
                return f"📋 活動【{activity}】剩餘名額：{row[0]} 人。"
            return "📋 活動包含：迎新晚會、程式工作坊、籃球友誼賽。"

        if action == "報名":
            if not activity: return "❌ 報名失敗：未指定活動。"
            if headcount < 1 or headcount > 3: return f"❌ 報名失敗：人數必須在 1~3 人之間。"

            cursor.execute('SELECT COUNT(DISTINCT activity_name) FROM registrations WHERE student_name = ? AND status = "成功"', (student_name,))
            row_cnt = cursor.fetchone()
            if row_cnt and row_cnt[0] >= 2: return f"❌ 報名失敗：每人最多同時報名 2 個活動！"

            cursor.execute('SELECT COUNT(*) FROM registrations WHERE student_name = ? AND activity_name = ? AND status = "成功"', (student_name, activity))
            row_dup = cursor.fetchone()
            if row_dup and row_dup[0] > 0: return f"❌ 報名失敗：您已重複報名【{activity}】！"

            cursor.execute('''
                SELECT a.quota - COALESCE(SUM(CASE WHEN r.status = '成功' THEN r.headcount ELSE 0 END), 0)
                FROM activities a LEFT JOIN registrations r ON a.name = r.activity_name AND r.status = '成功'
                WHERE a.name = ? GROUP BY a.id
            ''', (activity,))
            row_quota = cursor.fetchone()
            if not row_quota or row_quota[0] < headcount: return f"❌ 報名失敗：名額不足。"

            cursor.execute('INSERT INTO registrations (student_name, activity_name, headcount, status) VALUES (?, ?, ?, "成功")', (student_name, activity, headcount))
            conn.commit()
            return f"✅ 成功為【{student_name}】報名【{activity}】{headcount} 位！"

        if action == "取消":
            if not activity: return "❌ 請指定要取消的活動。"
            cursor.execute('SELECT id FROM registrations WHERE student_name = ? AND activity_name = ? AND status = "成功"', (student_name, activity))
            row_reg = cursor.fetchone()
            if not row_reg: return f"❌ 查無【{student_name}】報名【{activity}】的紀錄。"

            cursor.execute("UPDATE registrations SET status = '已取消' WHERE id = ?", (row_reg[0],))
            conn.commit()
            return f"✅ 已成功為【{student_name}】取消【{activity}】。"

        return "⚠️ 未知的操作。"
    finally:
        conn.close()

print("✅ 功能函式載入完成！")
````
